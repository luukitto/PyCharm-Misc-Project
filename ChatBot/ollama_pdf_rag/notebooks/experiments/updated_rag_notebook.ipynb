# Ollama PDF RAG Notebook

## Import Libraries


```python
# Imports
from langchain_community.document_loaders import UnstructuredPDFLoader, PyPDFLoader
from langchain_ollama import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain.prompts import ChatPromptTemplate, PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama.chat_models import ChatOllama
from langchain_core.runnables import RunnablePassthrough
from langchain.retrievers.multi_query import MultiQueryRetriever

# Suppress warnings
import warnings
warnings.filterwarnings('ignore')

# Jupyter-specific imports
from IPython.display import display, Markdown

# Set environment variable for protobuf
import os
os.environ["PROTOCOL_BUFFERS_PYTHON_IMPLEMENTATION"] = "python"
```

## Load PDF

```python
# Load PDF
local_path = "../../Georgia_codex.pdf"
if local_path:
    loader = PyPDFLoader(local_path)
    data = loader.load()
    print(f"PDF loaded successfully: {local_path}")
else:
    print("Upload a PDF file")
```

## Split text into chunks

```python
# Split text into chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = text_splitter.split_documents(data)
print(f"Text split into {len(chunks)} chunks")
```

## Create vector database

```python
# Create vector database
vector_db = Chroma.from_documents(
    documents=chunks,
    embedding=OllamaEmbeddings(model="nomic-embed-text"),
    collection_name="local-rag"
)
print("Vector database created successfully")
```

## Set up LLM and Retrieval

```python
# Set up LLM and retrieval
local_model = "llama3.2"  # or whichever model you prefer
llm = ChatOllama(model=local_model)
```

```python
# Query prompt template
QUERY_PROMPT = PromptTemplate(
    input_variables=["question"],
    template="""შენ ხარ AI ასისტენტი, რომელიც პასუხობს ქართულად დასმულ სამართლებრივ კითხვებს მხოლოდ საქართველოს სამოქალაქო კოდექსის საფუძველზე და ყოველთვის უთითებს წყაროს (მუხლი/მონაკვეთი).""",
)

# Set up retriever
retriever = MultiQueryRetriever.from_llm(
    vector_db.as_retriever(), 
    llm,
    prompt=QUERY_PROMPT
)
```

## Create chain

```python
# RAG prompt template
template = """უპასუხე შეკითხვას მხოლოდ არსებულ კონტენტზე დაყრდნობით:
{context}
კითხვა: {question}
"""

prompt = ChatPromptTemplate.from_template(template)
```

```python
# Create chain
chain = (
    {"context": retriever, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

## Chat with PDF

```python
def chat_with_pdf(question):
    """
    Chat with the PDF using the RAG chain.
    """
    return display(Markdown(chain.invoke(question)))
```

```python
# Example 1
chat_with_pdf("რა არის სამოქალაქო კანონმდეებლობა?")
```

```python
# Example 2
chat_with_pdf("What is the purpose of the scammer agent?")
```

```python
# Example 3
chat_with_pdf("Can you explain the case study highlighted in the document?")
```
